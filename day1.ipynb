```python
#data-types -

int_value = 30
print(int_value, type(int_value))

dec_value = 30.8
print(dec_value, type(dec_value))

complex_value = 2j+5
print(complex_value, type(complex_value))

value = -12.87
print(value, type(value))


#Operations

print("Absolute:", abs(value))
print("Rounded:", round(value))
```

```python
#Removing spaces and changing the letter case
raw_text = '        Pyhton Programming Is Very Useful'
clean_text = raw_text.strip()
print("Strip:-", clean_text)
print("Lower:-", clean_text.lower())
print("Upper:-", clean_text.upper())
print("Title:-", clean_text.title())
print("Capitalize:-",clean_text.capitalize())
print("Swap case:-",clean_text.swapcase())
```

```python
#Searching and checking string content.
string = "Believe in yourself, stay consistent and let your hard work create your success."
print("Find the word:-", string.find('rat'))
print("Count of character:-", string.lower().count("o"))
print("Is the word is uppercase?:-", "quick".isupper())
print("Ends with:-", string.endswith("Analysis")) 
print("Starts with:-", string.startswith("Believe"))
```

```python
#Replacing and Splitting and joining string.
updated_text = string.replace("fox", "Lion")
print(updated_text)
updated_words = updated_text.split()
print("Updated_words:", updated_words)
li = ['Kunal', 'Raghav', 'Kirtika', 'Lion', 'Rat - raghav']
print(type(li))
```
